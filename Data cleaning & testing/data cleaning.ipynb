```python
import pandas as pd
import numpy as np
print("pandas & numpy imported successfully")
```

```python
input_file = "Telco_Customer_Churn.xlsx"
df= pd.read_excel(input_file)
print(f"Dataset Shape: {df.shape[0]} rows and {df.shape[1]} columns")
df.head(3)
```

```python

df['Total Charges Clean'] = pd.to_numeric(df['Total Charges'], errors='coerce').fillna(0.0)


cols_to_drop = ['Count', 'Country', 'State']
df_cleaned = df.drop(columns=cols_to_drop)

print("Cleaned DataFrame columns count:", len(df_cleaned.columns))
```

```python

def categorize_tenure(months):
    if months <= 12:
        return '0-1 Year'
    elif months <= 24:
        return '1-2 Years'
    elif months <= 48:
        return '2-4 Years'
    else:
        return '4+ Years'

df_cleaned['Tenure Cohort'] = df_cleaned['Tenure Months'].apply(categorize_tenure)


addon_services = ['Online Security', 'Online Backup', 'Device Protection', 
                  'Tech Support', 'Streaming TV', 'Streaming Movies']
df_cleaned['Addon Services Count'] = df_cleaned[addon_services].apply(
    lambda row: sum(row == 'Yes'), axis=1
)


df_cleaned['Avg Monthly Spend Check'] = np.where(
    df_cleaned['Tenure Months'] > 0,
    df_cleaned['Total Charges Clean'] / df_cleaned['Tenure Months'],
    df_cleaned['Monthly Charges']
)


output_file = "cleaned_telco_churn.csv"
df_cleaned.to_csv(output_file, index=False)

print("=" * 50)
print(f"PHASE 1 COMPLETE: Saved {df_cleaned.shape[0]} rows to '{output_file}'!")
print("=" * 50)
df_cleaned[['CustomerID', 'Tenure Months', 'Tenure Cohort', 'Addon Services Count', 'Total Charges Clean']].head()
```
